# Notebook de pruebas

Aquí se realiza el request a la API, ingresando un input y recibiendo su predicción.

```python
import requests

url = "http://127.0.0.1:8000/predict"

oracion_1 = {
    "text": "there was an increment of 20%"
}

oracion_2 = {
    "text": "it was the worst for the business"
}

oracion_3 = {
    "text": "this was the best decision"
}

respuesta_1 = requests.post(url, json=oracion_1)
respuesta_2 = requests.post(url, json=oracion_2)
respuesta_3 = requests.post(url, json=oracion_3)

print("Oración 1:", respuesta_1.json())
print("Oración 2:", respuesta_2.json())
print("Oración 3:", respuesta_3.json())

# positive (2), neutral (1) or negative (0)
```
